```python
df.drop_duplicates(inplace = True)
```

```python
#replace dirty values
df.replace(['N/A', 'NULL', '', ' '], np.nan, inplace = True)
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()

text_columns = df.select_dtypes(include=['string', 'str']).columns

df[text_columns] = df[text_columns].astype('object')
df.info()
```

```python
# Remove extra spaces


df = df.apply(lambda x:x.str.strip() if x.dtype=="object" else x) 
```

```python
df.info()
```

```python
#convert text column
df["Customer_Name"] = df["Customer_Name"].str.title()
df["Gender"] = df["Gender"].str.title()
df["State"] = df["State"].str.title()
df["City"]= df["City"].str.title()
df["Subscription_Type"] = df["Subscription_Type"].str.title()
df["Contract_Type"] = df["Contract_Type"].str.title()
df["Churn"] = df["Churn"].str.title()


```

```python
df["Subscription_Type"].value_counts()
```

```python
df["Churn"].value_counts()
```

```python
#convert numeric column

cols = ["Age", "Tenure_Months", "Monthly_Charges", "Total_Charges"]

for c in cols:
    df[c]= pd.to_numeric(df[c],errors="coerce")
    
```

```python
df.info()
```

```python
df.describe()
```

```python
# removing invalid age

df = df[(df['Age']>=18)&(df['Age']<=100)]
```

```python
#Remove Negative Charges

df = df[df["Monthly_Charges"]>= 0]
df = df[df["Total_Charges"]>=0]
```

```python
df.shape
```

```python
df["Last_Interaction_Date"] = pd.to_datetime(
    df["Last_Interaction_Date"],
    errors="coerce"
)
```

```python
#Fill Missing Values
df["Tech_Support"] = df["Tech_Support"].fillna("No")

df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")

df["Monthly_Charges"] = df["Monthly_Charges"].fillna(
    df["Monthly_Charges"].mean())

df["Tenure_Months"] = df["Tenure_Months"].fillna(
    df["Tenure_Months"].mean())

df["Age"] = df["Age"].fillna(
    df["Age"].mean())
```

```python
# Feature Engineering
#Customer Lifetime Value
df['Customer_Value'] = df["Monthly_Charges"]*df["Tenure_Months"]
```

```python
df.head()
```

```python
#monthly revenue

df["Monthly_Revenue"]=df["Monthly_Charges"]
```

```python
#Tenure Group "0-12", "13-24", "25-48", "49-72"
bins =[0,12,24,48,72]
label= labels = ["0-12 month", "13-24 month", "25-48 month", "49-72 month"]  
df["Tenure_Group"]= pd.cut(df["Tenure_Months"], bins=bins, labels = label)

```

```python
df.head()
```

```python
# senior citizen flag

df["Senior_Flag"]=np.where(df["Age"]>=60, "Senior", "Adult")
```

```python
df.head()
```

```python
# churn flag

df["Churn_Flag"]= df["Churn"].map({"Yes":1,"No":0})
```

```python
df.describe()
```

```python
df["Last_Interaction_Date"] = df["Last_Interaction_Date"].fillna(
    df["Last_Interaction_Date"].median()
)
```

```python
df.info()
```

```python
df.describe()
```

```python
# Export clean dataset
df.to_csv(
    r"C:\Users\Nikita Dey\Downloads\Data Analysis Project\Data Analysis Project\Clean_Churn_Data.csv",
    index=False
)
```
